```python
import requests

url = "https://api.erg.ic.ac.uk/AirQuality/Information/MonitoringSites/GroupName=London/Json"
response = requests.get(url)

print(response.status_code)
data = response.json()
```

```python
type(data)
```

```python
import json

with open("data/monitoring_sites_london.json", "w") as f:
    json.dump(data, f, indent=2)
```

```python
import duckdb


duckdb.sql("""SELECT DISTINCT areas."@LocalAuthorityName"
FROM(
    SELECT unnest(Sites.Site) AS areas
    FROM read_json_auto('data/monitoring_sites_london.json'))""").df()
```

```python
duckdb.sql("""
SELECT areas."@LocalAuthorityName", COUNT(*) AS n_sites
FROM (
    SELECT unnest(Sites.Site) AS areas
    FROM read_json_auto('data/monitoring_sites_london.json')
)
GROUP BY areas."@LocalAuthorityName"
ORDER BY n_sites DESC
""").df()
```

```python
url = "https://api.erg.ic.ac.uk/AirQuality/Information/MonitoringSiteSpecies/GroupName=London/Json"
response = requests.get(url)

print(response.status_code)
data = response.json()

with open("data/area_species_tracked.json", "w") as f:
    json.dump(data, f, indent=2)

duckdb.sql("""
SELECT area_species.*
FROM (
    SELECT unnest(Sites.Site) AS area_species
    FROM read_json_auto('data/area_species_tracked.json')
)
WHERE area_species."@LocalAuthorityName" IN ('Croydon', 'Merton', 'Sutton')
AND (area_species."@DateClosed" IS NULL OR area_species."@DateClosed" = '')
""").df()
```
